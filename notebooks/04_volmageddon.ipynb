# 04 · Volmageddon, 5 February 2018 (descriptive)
On 5 Feb 2018 the VIX closed at 37.32 after opening at 18.44, and inverse-VIX products (XIV, SVXY) lost ~97% overnight (Augustin, Cheng & Van den Bergen, 2021).

**This notebook is descriptive.** It is not part of the thesis backtest, which starts in April 2019 because 2018 is consumed by the first training window. No P&L is simulated here.

```python
from pathlib import Path
import pandas as pd
from vrp import data, features, models, backtest, plots

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
RAW = ROOT / "data" / "raw"
pd.set_option("display.precision", 4)
```

```python
raw = data.load_raw(RAW)
feats = features.build_features(raw["underlying"], raw["chain"], raw["macro"])
fig = plots.volmageddon(feats)
```

```python
w = feats[(feats.date >= "2018-01-22") & (feats.date <= "2018-02-16")]
w[["date", "mfiv", "rv_yz", "rv_yz_forward", "target"]]
```

In the two weeks before the shock, 30-day implied vol sat around 11% while the S&P 500 went on to realise ~21–23% over the following month: the realised VRP was about −10 vol points. A short-variance position opened in late January was priced for a calm that did not last. Once the VIX spiked, implied vol overshot what was subsequently realised and the premium turned sharply positive again: the classic pattern of a premium that pays most of the time and gives back in bursts.
